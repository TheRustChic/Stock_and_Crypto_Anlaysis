```python
import pandas as pd

df_tesla = pd.read_excel("sheet.xlsx", sheet_name="Tesla")

```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Calculate daily returns
df_tesla['Return'] = df_tesla['close'].pct_change()

# 7-day rolling volatility
df_tesla['Volatility'] = df_tesla['Return'].rolling(7).std()


# Convert 'date' column to datetime (just in case it's not already)
df_tesla['Date'] = pd.to_datetime(df_tesla['Date'])

# Plot
plt.figure(figsize=(12, 5))
plt.plot(df_tesla['Date'], df_tesla['volatility'], label='7-Day Volatility')

# Format date axis
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
plt.gca().xaxis.set_major_locator(mdates.MonthLocator())  # Change to WeekLocator() if you want more granularity

plt.title("Tesla Stock Volatility")
plt.xlabel("Date")
plt.ylabel("Volatility")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.savefig("tesla_volatility.png", dpi=300)  # <-- Save before showing
plt.show()



```

```python
import pandas as pd
import numpy as np

# Assuming df_tesla is the DataFrame containing the Tesla data

# 1. Calculate Daily Returns
df_tesla['Return'] = df_tesla['close'].pct_change()

# 2. Calculate Cumulative Return
df_tesla['Cumulative Return'] = (1 + df_tesla['Return']).cumprod() - 1

# 3. Calculate Volatility (Standard Deviation of Daily Returns)
volatility = df_tesla['Return'].std()

# 4. Calculate Average Daily Return
average_daily_return = df_tesla['Return'].mean()

# 5. Calculate 30-Day Simple Moving Average (SMA)
df_tesla['30-Day SMA'] = df_tesla['close'].rolling(window=30).mean()

# 6. Get the Latest 30-Day SMA Value
latest_sma = df_tesla['30-Day SMA'].iloc[-1]  # Last value in the 30-Day SMA column

# 7. Total Return (final value in 'Cumulative Return' column)
total_return = df_tesla['Cumulative Return'].iloc[-1]

# Prepare the summary table
summary = pd.DataFrame({
    'Metric': ['Total Return', 'Volatility (Daily)', 'Average Daily Return', 'Latest 30-Day SMA'],
    'Value': [f'{total_return*100:.2f}%', f'{volatility*100:.2f}%', f'{average_daily_return*100:.2f}%', f'${latest_sma:.2f}']
})

# Print the summary table
print(summary)





```
